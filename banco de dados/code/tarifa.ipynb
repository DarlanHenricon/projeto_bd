```python
import pandas as pd
```

```python
caminho = 'tarifas-homologadas-distribuidoras-energia-eletrica.csv'
```

```python
df = pd.read_csv(caminho, sep=';', encoding='latin-1')
```

```python
print(df.head(5))
```

```python
# Filtrar tarifas residenciais convencionais
df = df[
    (df['DscBaseTarifaria'] == 'Tarifa de Aplicação') &
    (df['DscSubGrupo'] == 'B1') &
    (df['DscClasse'] == 'Residencial') &
    (df['DscSubClasse'] == 'Residencial') &
    (df['DscModalidadeTarifaria'] == 'Convencional') &
    (df['DscUnidadeTerciaria'] == 'MWh') &
    (df['NomPostoTarifario'] == 'Não se aplica')
].copy()
```

```python
for coluna in ['VlrTUSD', 'VlrTE']:
    df[coluna] = pd.to_numeric(
        df[coluna].astype(str)
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False),
        errors='coerce'
    )
```

```python
# Criar tarifa total
df['Tarifa_Total'] = (
    df['VlrTUSD'] + df['VlrTE']
)
```
