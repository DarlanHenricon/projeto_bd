```python
tarifa_kwh = 0.74

analise_regional = (
    df.groupby(['Regiao', 'Classe'])
    .agg(
        consumo_medio_gwh=('Consumo', 'mean'),
        consumo_total_gwh=('Consumo', 'sum'),
        consumidores_media=('Consumidores', 'mean'),
        registros=('Consumo', 'count')
    )
)

# Converter MWh para GWh
analise_regional['consumo_medio_gwh'] /= 1000
analise_regional['consumo_total_gwh'] /= 1000

# Calcular receita em bilhões de reais
analise_regional['receita_bilhoes'] = (
    analise_regional['consumo_total_gwh'] * tarifa_kwh * 0.001
)

# Ordenar pelo consumo médio (maior para menor)
analise_regional = analise_regional.sort_values(
    by='consumo_medio_gwh',
    ascending=False
)

# Arredondar valores numéricos
analise_regional = analise_regional.round(2)

# Formatar consumo com unidade GWh
for coluna in ['consumo_medio_gwh', 'consumo_total_gwh']:
    analise_regional[coluna] = analise_regional[coluna].apply(
        lambda x: f'{x:,.2f} GWh'
        .replace(',', 'X')
        .replace('.', ',')
        .replace('X', '.')
    )

# Formatar receita
analise_regional['receita_bilhoes'] = analise_regional['receita_bilhoes'].apply(
    lambda x: f'R$ {x:,.2f} bilhões'
    .replace(',', 'X')
    .replace('.', ',')
    .replace('X', '.')
)

display(analise_regional)
```
